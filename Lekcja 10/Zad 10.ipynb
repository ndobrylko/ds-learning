Dodaj bonus do każdego egzaminu: [5, 10, 5, 10]
Następnie pomnóż wyniki każdego studenta przez mnożnik: [1.0, 1.1, 0.95, 1.15, 1.0]
Wymagania:<br>
 - Użyj broadcastingu (bez pętli)
- Wypisz wyniki przed i po



```python
import numpy as np
```

```python
exams = np.array([
    [78, 85, 92, 88],
    [65, 72, 68, 70],
    [90, 88, 95, 92],
    [55, 60, 58, 62],
    [82, 79, 85, 80]
])

bonus = np.array([5, 10, 5, 10])
mnożnik = np.array([1.0, 1.1, 0.95, 1.15, 1.0])

print(f"Wyniki przed: {exams}")
```

```python
wyniki_po = (exams + bonus) * mnożnik[:, np.newaxis]
#np.newaxis tworzy (5,1) (wcześniej było (5,)
#mnożnik[:] - wszystkie elementy
print(f"Wyniki po: {wyniki_po}")
```
