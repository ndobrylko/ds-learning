Dana tablica 1D: temps = np.array([15, 18, 22, 25, 23, 20, 17, 14, 16, 19, 21, 24])
Wykonaj:
(proste)

Wypisz temperatury z dni parzystych (0, 2, 4, ...)
Wypisz ostatnie 5 temperatur
Wypisz temperatury w odwrotnej kolejności
Znajdź temperatury większe niż 20

```python
import numpy as np

temps = np.array([15, 18, 22, 25, 23, 20, 17, 14, 16, 19, 21, 24])

dni_parzyste = temps[::2]
ostatnie_5 = temps[-5:]
odwrotnie = temps[::-1]
powyzej_20 = temps[temps > 20]

print("Dni parzyste:", dni_parzyste)
print("Ostatnie_5: ", ostatnie_5)
print("Powyzej_20: ", powyzej_20)
print("Odwrotnie: ", odwrotnie)
```
