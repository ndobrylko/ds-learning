Dana tablica reprezentująca sprzedaż 4 produktów w 7 dni:
sales = np.random.randint(10, 100, size=(4, 7))
Oblicz i wypisz:
(proste)
Średnią sprzedaż per produkt (axis=1)
Średnią sprzedaż per dzień (axis=0)
Dzień z najwyższą sprzedażą (np.argmax)
Produkt z najniższą średnią sprzedażą

```python
import numpy as np

sales = np.random.randint(10,100, size=(4,7))

srednia_produkt = np.mean(sales, axis=1)
srednia_dzien = np.mean(sales, axis=0)
dzien_max = np.argmax(np.sum(sales, axis=0))
produkt_min = np.argmin(srednia_produkt)

print("Średnia sprzedaz per produkt:", srednia_produkt)
print("Sr sprzedaż per dzień:", srednia_dzien)
print("Dzień z najw sprzedażą (indeks):", dzien_max)
print("Produkt z najniższą średnią sprzedażą (indeks):", produkt_min)
```
