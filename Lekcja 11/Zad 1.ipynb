```python
data = {
'Produkt': ['Laptop', 'Mysz', 'Klawiatura', 'Monitor', 'Słuchawki'],
'Cena': [3500, 80, 250, 1200, 350],
'Ilość na stanie': [45, 320, 150, 78, 210],
'Kategoria': [3.5, 4.2, 4.7, 4.4, 4.3]
}

df = pd.DataFrame(data)
df

```

```python
df["Wartość całkowita"] = df["Cena"] * df["Ilość na stanie"]
```
