Sprawdź które kolumny w Titanicu mają brakujące dane.
Wymagania:
Policz NaN w każdej kolumnie
✏ Zadania średnie (9-12)
(proste)
Oblicz procent NaN
Wyświetl tylko kolumny z >10% braków

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
```

```python
nan_count = df.isna().sum()
```

```python
nan_percent = (nan_count / len(df)) * 100
nan_summary = pd.DataFrame({"Liczba_Nan": nan_count, "Liczba_Nan_percent": nan_percent})
```

```python
nan_summary[nan_summary["Liczba_Nan_percent"] > 10]
```
