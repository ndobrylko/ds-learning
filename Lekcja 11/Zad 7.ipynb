Posortuj pasażerów Titanica po wieku (od najmłodszego do najstarszego).
Wymagania:<br>
Użyj .sort_values()<br>
Wyświetl 10 najmłodszych<br>
Wyświetl 10 najstarszych<br>

```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)
```

```python
titanic_sorted = titanic.sort_values(by='Age', ascending=True)
print(f"10 najmłodszych pasażerów: {titanic_sorted.head(10)}")
```

```python
print(f"10 najstarszych pasażerów: {titanic_sorted.tail(10)}")
```
