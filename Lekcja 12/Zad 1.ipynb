Wczytaj dataset Titanic i oblicz średni wiek pasażerów według klasy ( Pclass ) i płci ( Sex ).
Dataset: Titanic
(https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv) <br>
Wymagania:<br>
Użyj .groupby() z dwoma kolumnami<br>
Oblicz średnią wieku ( .mean() )<br>
Wyświetl wyniki w formie czytelnej tabeli<br>
Oczekiwany rezultat:<br>
Tabela ze średnim wiekiem dla 6 kombinacji (3 klasy × 2 płcie)<br>

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head() #5 pierwszych wierszy
df.info() #dane techniczne, l. wierszy, kolumn, nazwy kolumn, typ danych kazdej kolumny, NaN(brakujace wartosci)
```

```python
result = df.groupby(['Pclass', 'Sex'])['Age'].mean().reset_index()
#groupby(['Pclass', 'Sex']) Grupuje dane po klasie i płci
#['Age']	Wybiera kolumnę z wiekiem
#.mean()	Liczy średnią wieku w każdej grupie
#.reset_index()	Zamienia indeksy grupowania na normalne kolumny

print(result.round(2))
```
