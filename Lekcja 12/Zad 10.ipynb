Napisz własną funkcję, która oblicza zakres międzykwartylowy (IQR = Q3 - Q1) i zastosuj ją
do grupy.<br>
Dataset: California Housing<br>
Wymagania:<br>
Napisz funkcję iqr_func(x) obliczającą IQR<br>
Zastosuj ją do MedHouseVal według regionów<br>
Porównaj IQR między regionami<br>
Oczekiwany rezultat:<br>
Tabela z IQR dla każdego regionu<br>

```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

housing = fetch_california_housing(as_frame=True)
df = pd.concat([housing.data, housing.target.rename('MedHouseVal')], axis=1)
#axis=1 - łaczenie kolumnami

def iqr_func(x):
    Q1 = x.quantile(0.25)
    Q3 = x.quantile(0.75)
    return Q3 - Q1

df['Region'] = pd.qcut(df['Latitude'], 5, labels=False) #pd.qcut dzieli dane na grupy (5)
# Jeśli ustawimy labels=False, Pandas zamiast nazw przedziałów zwróci numeryczne indeksy grup (0,1,2,3,4,5)

iqr_by_region = df.groupby('Region')['MedHouseVal'].agg(iqr_func).reset_index()
#reset_index() - Zamienia indeks grupy (Region) w normalną kolumnę w DataFrame.
iqr_by_region
```
