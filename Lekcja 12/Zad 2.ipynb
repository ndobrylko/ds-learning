Dla datasetu Titanic policz, ile pasażerów było w każdej klasie i czy przeżyli ( Survived ).
Dataset: Titanic
Wymagania:<br>
Grupuj według Pclass i Survived<br>
Użyj .size() lub .count()<br>
Przedstaw wyniki jako tabelę pivot<br>
Oczekiwany rezultat:
Tabela pokazująca liczbę pasażerów dla każdej kombinacji klasy i statusu przeżycia

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
grouped = df.groupby(['Pclass', 'Survived']).size().reset_index(name='Count')
#groupby(['Pclass', 'Survived'])	grupujemy po klasie i statusie przeżycia
#.size()	liczy wszystkie wiersze w grupie / .count() liczy tylko nie-NaN w wybranej kolumnie
#reset_index()	zamienia indeks na kolumny
#name='Count'	nazywa kolumnę z liczbą pasażerów
```

```python
pivot_table = grouped.pivot(
    index='Pclass',
    columns='Survived',
    values='Count'
)
print(pivot_table)
```

```python
df.groupby(['Pclass', 'Survived']).size().unstack()
#unstack() robi dokładnie to samo co pivot()
```
