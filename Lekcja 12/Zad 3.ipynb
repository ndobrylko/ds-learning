```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
sales = pd.DataFrame({
'Region': np.random.choice(['A', 'B', 'C'], 100),
'Sprzedaż': np.random.randint(50, 500, 100)
})
```

```python
sales.head()
```

```python
result = sales.groupby('Region')['Sprzedaż'].agg(['min', 'max', 'mean', 'std'])
#groupby('Region')	dzieli dane na regiony A, B, C
#['Sprzedaż']	wybiera kolumnę liczbową
#.agg([...])	liczy wiele statystyk naraz
#min, max, mean, std	minimum, maksimum, średnia, odchylenie
result = result.round(2)
print(result)
```
