Dodaj do datasetu Titanic nową kolumnę Avg_Fare_Pclass , która zawiera średnią opłatę
( Fare ) dla każdej klasy.<br>
Dataset: Titanic<br>
Wymagania:<br>
Użyj .groupby().transform('mean')<br>
Dodaj kolumnę do oryginalnego DataFrame<br>
Wyświetl pierwsze 10 wierszy z kolumnami: Pclass , Fare , Avg_Fare_Pclass<br>
Oczekiwany rezultat:<br>
DataFrame z nową kolumną, każdy wiersz ma średnią swojej klasy<br>

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df['Avg_Fare_Pclass'] = df.groupby('Pclass')['Fare'].transform('mean')
df[['Pclass', 'Fare', 'Avg_Fare_Pclass']].head(10)
```
