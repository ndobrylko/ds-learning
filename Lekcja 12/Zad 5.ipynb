Podziel wiek pasażerów Titanica na 4 kategorie: Dziecko (0-12), Nastolatek (13-18),
Dorosły (19-60), Senior (60+) Dataset: Titanic
Wymagania: <br>
Użyj pd.cut() z własnymi przedziałami<br>
Oblicz survival rate dla każdej kategorii wiekowej<br>
Przedstaw wyniki jako wykres słupkowy<br>
Oczekiwany rezultat:<br>
Wykres showing survival rate według kategorii wiekowych<br>

```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)


bins = [0, 12, 18, 60, float('inf')]
labels = ['Dziecko', 'Nastolatek', 'Dorosły', 'Senior']
#float('inf') = nieskończoność (wszystko powyżej 60)

df['Age_Category'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels
)
# Utworzenie nowej kolumny z kategorią wieku

survival_rate = df.groupby('Age_Category', observed=True)['Survived'].mean()


#Wykres słupkowy:
!pip install matplotlib

import matplotlib.pyplot as plt

plt.figure() #Tworzy nową „figurę” (okno wykresu), czyli kontener dla wykresu.
survival_rate.plot(kind='bar') #Tworzy wykres słupkowy (bar) z danych Pandas (survival_rate – Series).
plt.ylabel('Survival Rate')
plt.xlabel('Kategoria wiekowa')
plt.title('Survival Rate według kategorii wiekowych')
plt.show()
```
