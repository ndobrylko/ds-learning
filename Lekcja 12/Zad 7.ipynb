Dla datasetu California Housing znajdź 5 regionów z najwyższą średnią ceną domu.
Dataset: California Housing (sklearn)<br>
Wymagania:<br>
Stwórz regiony na podstawie Latitude/Longitude (podziel na siatkę)<br>
Oblicz średnią cenę według regionu<br>
Posortuj malejąco i wyświetl top 5<br>
Oczekiwany rezultat:<br>
Lista 5 regionów z najwyższą średnią ceną<br>

```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

housing = fetch_california_housing(as_frame=True)
df = pd.concat([housing.data, housing.target.rename('MedHouseVal')], axis=1)

#pd.qcut dzieli dane na grupy (5)
# Jeśli ustawimy labels=False, Pandas zamiast nazw przedziałów zwróci numeryczne indeksy grup (0,1,2,3,4,5)
df['Lat_Region'] = pd.qcut(df['Latitude'], 5, labels=False)
df['Lon_Region'] = pd.qcut(df['Longitude'], 5, labels=False)

# polaczenie w jeden region jako tuple
df['Region'] = list(zip(df['Lat_Region'], df['Lon_Region']))

mean_price_by_region = df.groupby('Region')['MedHouseVal'].mean()
top5_regions = mean_price_by_region.sort_values(ascending=False).head(5)
top5_regions

```
