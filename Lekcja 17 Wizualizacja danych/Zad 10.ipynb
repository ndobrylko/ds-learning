Utwórz wykres ceny akcji w czasie z:
- Główną linią ceny<br>
- Średnią kroczącą 7-dniową (MA7)<br>
- Średnią kroczącą 21-dniową (MA21)<br>
- Zacienionym obszarem między MA7 a MA21<br>

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

dates = pd.date_range('2025-01-01', periods=100)
price = 100 + np.cumsum(np.random.randn(100))

df = pd.DataFrame({
    'date': dates,
    'price': price
})

#pusty wykres:
fig = make_subplots(
    rows=1,
    cols=1,
    shared_xaxes=True, #wspólna oś X
    subplot_titles=('Cena akcji',)
)

fig.add_trace(
    go.Scatter(
        x=dates,
        y=price,
        mode='lines',
        name='Główna linia ceny',
        line=dict(color='green', width=2)
    ),
    row=1,
    col=1
)

#obliczanie ma7
ma7 = pd.Series(price).rolling(window=7).mean() #tworzy 7dniowe okno kroczące i liczy śr w każdym oknie

#dodanie ma7 do wykresu
fig.add_trace(
go.Scatter(
    x=dates,
    y=ma7,
    mode='lines',
    name='MA7',
    line=dict(color='orange', width=1.5)),
row=1, col=1
)

#obliczanie ma21
ma21 = pd.Series(price).rolling(window=21).mean()

#dodanie ma21 do wykresu
fig.add_trace(
go.Scatter(x=dates, y=ma21, mode='lines',
name='MA21', line=dict(color='blue', width=1.5)),
row=1, col=1
)

#dodanie niewidocznej linii m7; fill='tonexty' wypełnia obszar nie do konkretnych danych a do linijki wcześniej
fig.add_trace(
    go.Scatter(
        x=dates,
        y=ma7,
        line=dict(width=0),
        showlegend=False
    ),
    row=1, col=1
)

#zacieniony obszar
fig.add_trace(
    go.Scatter(
        x=dates,
        y=ma21,
        fill='tonexty',
        fillcolor='rgba(255,165,0,0.2)',
        line=dict(width=0), #brak widocznej linii
        name='Obszar MA7–MA21'
    ),
    row=1, col=1
)
```
