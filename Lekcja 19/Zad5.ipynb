Dla funkcji f(x) = x² uruchom Gradient Descent z różnymi learning rates.
Wymagania:<br>
-Przetestuj: η = 0.001, 0.01, 0.1, 0.5, 1.0, 1.5<br>
-Punkt startowy: x = 5<br>
-30 iteracji<br>
-Określ, które η działa najlepiej<br>
Oczekiwany rezultat:<br>
-Wykres krzywych uczenia dla każdego η<br>
-Wnioski o optymalnym learning rate<br>

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return x ** 2

#pochodna niezbedna do wzoru GD
def gradient(x):
    return 2 * x

def run_gd(start_x, learning_rate, num_iterations):
    """Uruchomienie Gradient Descent z określonym learning rate."""
    x = start_x
    history = [x]

    for _ in range(num_iterations):
        x = x - learning_rate * gradient(x)
        history.append(x)

    return history

learning_rates = [0.001, 0.01, 0.1, 0.5, 1.0, 1.5]
start_x = 5
iterations = 30

results = {}

for eta in learning_rates:
    history = run_gd(start_x, eta, iterations)
    results[eta] = history

for eta in learning_rates:
    plt.plot(results[eta], label=f"eta={eta}")

plt.xlabel("Iteracja")
plt.ylabel("x")
plt.title("Gradient Descent dla różnych learning rates")
plt.legend()
plt.show()

```

Learning rate 1,5 zasłaniają pozostałe krzywe, wartości są zbyt duże i wykres został do nich dostosowany. Reszta krzywych oscyluje wokół 0.

```python
learning_rates_max_1 = [0.001, 0.01, 0.1, 0.5, 1.0]
start_x = 5
iterations = 30

results = {}

for eta in learning_rates_max_1:
    history = run_gd(start_x, eta, iterations)
    results[eta] = history

for eta in learning_rates_max_1:
    plt.plot(results[eta], label=f"eta={eta}")

plt.xlabel("Iteracja")
plt.ylabel("x")
plt.title("Gradient Descent dla różnych learning rates (max_0.5)")
plt.legend()
plt.show()
```

Optymalne learning rate to 0,1, działa poprawnie i stabilnie. Jednak najszybszą zbieżność zapewnia 0.5 minimalizuje funkcję błędu w najmniejszej liczbie iteracji.
