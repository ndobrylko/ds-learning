```python
import math

learning_rate = 0.01
batch_size = 64
epochs = 100
dropout = 0.5

# learning_rate
if 0 < learning_rate <= 1:
    print("learning_rate: OK")
else:
    print("learning_rate: BŁĄD – powinien być w (0,1]")

# batch_size
if batch_size > 0 and (batch_size & (batch_size - 1)) == 0:
    print("batch_size: OK")
else:
    print("batch_size: BŁĄD – powinien być potęgą 2")

# epochs
if 0 < epochs < 1000:
    print("epochs: OK")
else:
    print("epochs: BŁĄD – powinien być >0 i <1000")

# dropout
if 0 <= dropout <= 1:
    print("dropout: OK")
else:
    print("dropout: BŁĄD – powinien być w [0,1]")
```
