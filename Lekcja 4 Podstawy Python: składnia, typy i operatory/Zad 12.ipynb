```python
# Dane
n_samples = 50000
batch_size = 128
n_epochs = 50
time_per_batch = 0.15  # sekundy

# Liczba batchy na epokę
batches_full = n_samples // batch_size
remaining = n_samples % batch_size
batches_per_epoch = batches_full + 1 if remaining > 0 else batches_full

# Całkowity czas w sekundach
total_seconds = batches_per_epoch * time_per_batch * n_epochs

# Konwersja na godziny, minuty, sekundy
hours = int(total_seconds // 3600)
minutes = int((total_seconds % 3600) // 60)
seconds = total_seconds % 60

# Wyświetlenie wyniku
print(f"Szacowany czas: {hours} h {minutes} min {seconds:.0f} s")
```
