```python
values = [75, 50, 100, 62.5]
min_val = 50
max_val = 100

def normalize(value, min_val, max_val):
    return (value - min_val) / (max_val - min_val)

print(f"{'Original':<10} | {'Normalized':>10}")
print("-"*25)
```

```python
for value in values:
    norm = normalize(value, min_val, max_val)
    print(f"{value:<10} | {norm:>10.2f}")
```

```python
print(f"{'Original':<10} | {'Normalized':>10}")
print("-"*25)
for value in values:
    norm = normalize(value, min_val, max_val)
    print(f"{value:<10} | {norm:>10.2f}")
```
