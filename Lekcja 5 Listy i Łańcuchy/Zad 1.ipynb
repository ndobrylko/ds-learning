Stwórz listę zawierającą pomiary temperatury z 7 dni: [22.5, 23.0, 21.5, 24.0, 23.5,
22.0, 23.0] . Wyświetl:
(proste)
Temperaturę z pierwszego dnia
Temperaturę z ostatniego dnia
Temperatury z trzech środkowych dni (slicing)

```python
temperatury = [22.5, 23.0, 21.5, 24.0, 23.5, 22.0, 23.0]
print(f"Temp z pierwszego dnia: {temperatury [0]} st C")
print(f"Temp z ost dnia: {temperatury [-1]} st C")
print(f"Temp z trzech środkowych dni: {temperatury [2:5]} st C")
```
