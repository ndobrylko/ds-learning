Oblicz łączną kwotę dla każdej kategorii. Wynik zapisz w słowniku (użyj pętli i warunków).

```python
transactions = [
    ('food', 50),
    ('transport', 20),
    ('food', 30),
    ('entertainment', 40),
    ('transport', 15),
    ('food', 25)
]
#tworzymy pusty słownik
suma = {}

#przechodzimy pętlą po transakcjach i sprawdzamy czy kategoria już jest w słowniku - jeśli nie dodajemy i przypisujemy wartośc, jeśli tak dopisujemy tylko wartość


# pętla bierze kazdy element listy food[0] 50[1]; transport [0], 20[1] itd
for trans in transactions:
    kategoria = trans[0]
    kwota = trans[1]

    if kategoria in suma:
        suma[kategoria] += kwota
    else:
        suma[kategoria] = kwota

print(suma)
```

```python
#rozwiązanie z dict.get()
transactions = [
    ('food', 50),
    ('transport', 20),
    ('food', 30),
    ('entertainment', 40),
    ('transport', 15),
    ('food', 25)
]

suma = {}

#automatyczne rozpakowanie krotki (kategoria, kwota)
for kategoria, kwota in transactions:
    suma[kategoria] = suma.get(kategoria, 0) + kwota

print(suma)
```
