Zacznij od pustej listy results = [] . Dodaj do niej 5 wartości accuracy: 0.75, 0.82, 0.79,
0.85, 0.88. Następnie:
Wyświetl długość listy
Wyświetl średnią wartość
Wyświetl maksymalną wartość

```python
results = []
#Dodawanie przez extend (cała lista/ append to 1 element)
results.extend([0.75, 0.82, 0.79,
0.85, 0.88])
#Dług listy
print(f"Liczba wartości: {len(results)}")
#Śr wartość
srednia_wartosc = sum(results) / len(results)
print(f"Średnia wartość: {srednia_wartosc}")
#Max wartosc
max_wartosc = max(results)
print(f"Maksymalna wartość: {max_wartosc}")

```
