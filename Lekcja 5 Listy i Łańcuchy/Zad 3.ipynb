Masz listę wyników: [0.78, 0.92, 0.85, 0.88, 0.81] .
Posortuj ją malejąco i wyświetl 3 najlepsze wyniki

```python
results = [0.78, 0.92, 0.85, 0.88, 0.81]
#Sortowanie
results.sort()
print(results)
#Sortowanie malejąco
results.reverse()
print(results)
#3 najlepsze wyniki
print(f"3 najlepsze wyniki: {results[0:3]}")
```

```python
#szybsze rozwiązanie
top3 = sorted(results, reverse=True)[:3]
print(top3)
```
