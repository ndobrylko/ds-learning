Masz listę nazw kolumn: [" Customer ID ", "TOTAL AMOUNT", " date "] . Wyczyść je
zgodnie z konwencją:
(proste)
Usuń białe znaki (strip)
Zamień na małe litery (lower)
Zamień spacje na podkreślniki (replace)

```python
columns = [" Customer ID ", "TOTAL AMOUNT", " date "]
#nie działa powszechny błąd!!! to jest tylko zmienna, nie wpływamy na rzeczywistą listę!!
for column in columns:
    column = column.strip()
    column = column.lower()
    column = column.replace(' ', '_')
print(columns)

#list comprehension
columns = [col.strip().lower().replace(' ', '_') for col in columns]
print(columns)

```
