Zadanie 5 – Parsowanie CSV
Masz string CSV: "Alice,28,Data Scientist,75000" .
Użyj split(',') aby wyciągnąć:
Imię
Wiek (jako int)
Zawód
Pensję (jako int)
Wyświetl te informacje w formacie: "Alice ma 28 lat, pracuje jako Data Scientist i zarabia $75000"
(proste)


```python
#csv czyli comma-separated value
csv_line = "Alice,28,Data Scientist,75000"
#split nam dzieli string i zawsze zwraca stringi (nawet jesli w CSV były liczby
fields = csv_line.split(',')
print(fields)

#rozpakowujemy zmienne
imie, wiek, zawod, pensja = fields

#konwersja na int
wiek = int(wiek)
pensja = int(pensja)

print(f"{imie} ma {wiek} lat, pracuje jako {zawod} i zarabia {pensja}$")
```
