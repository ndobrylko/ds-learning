Masz dane: model_name = "XGBoost" , accuracy = 0.8934 , time = 45.67 . Użyj f-string aby stworzyć raport:
Model: XGBoost
Accuracy: 89.34%
Training time: 45.67s

```python
model_name = "XGBoost"
accuracy = 0.8934
time = 45.67

print(f"""Model: {model_name}
Accuracy: {accuracy:.2%}
Training time: {time:.2f}""")

# """ - wielolinijkowy string

#f-string - pozwala wstawić zmienną w string bez łączenia z +, pozwala też formatować liczby w locie

#.2% → dwie cyfry po przecinku i przekształcenie ułamka w procent

#.2f → float z dokładnością do 2 miejsc po przecinku

#W f-stringu zapis w {} ma strukturę:
{zmienna : format_spec}
zmienna → np. accuracy
: → separator, który mówi Pythonowi: „zaraz podam format”
ormat_spec → instrukcja formatująca np. .2% lub .2f
```
