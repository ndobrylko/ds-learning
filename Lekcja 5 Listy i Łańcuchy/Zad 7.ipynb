Masz listę predykcji: ['spam', 'ham', 'spam', 'spam', 'ham', 'ham', 'spam'] .
Policz ile jest 'spam' i ile 'ham' używając metody count() .

```python
predykcje = ['spam', 'ham', 'spam', 'spam', 'ham', 'ham', 'spam']
ilosc_spam = predykcje.count('spam')
print(ilosc_spam)

ilosc_ham = predykcje.count('ham')
print(ilosc_ham)
```
