Masz listę pomiarów: [23, 24, 22, 150, 23, 25, -10, 24, 23] . Outlierem jest wartość
poza zakresem [20, 30]. Stwórz nową listę bez outlierów używając list comprehension.

```python
pomiary = [23, 24, 22, 150, 23, 25, -10, 24, 23]

#[nowy_element FOR element IN lista IF warunek]
czyste_pomiary = [p for p in pomiary if 20 <= p <= 30]
print(czyste_pomiary)
```

```python
scores = [55, 78, 102, 90, 67, -5, 88, 73, 150]
clean_scores = [s for s in scores if 60 <= s <= 100]
print(clean_scores)
```
