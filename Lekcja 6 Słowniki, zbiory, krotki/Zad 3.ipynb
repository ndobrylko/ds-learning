Masz listę słów: ['python', 'java', 'python', 'javascript', 'python', 'java',
'c++', 'python'] . Policz ile razy każde słowo występuje używając słownika.
Wymagania:
Oczekiwany wynik:
(proste)
Użyj pętli i metody .get()
Wypisz wyniki
Słownik z częstościami: {'python': 4, 'java': 2, ...}

```python
jezyki = ['python', 'java', 'python', 'javascript', 'python', 'java', 'c++', 'python']

czestosci = {}
for jezyk in jezyki:
    czestosci[jezyk] = czestosci.get(jezyk, 0)+1

print(czestosci)
```
