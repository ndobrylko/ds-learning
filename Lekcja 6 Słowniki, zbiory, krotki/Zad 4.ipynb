Masz dwa zbiory: A = {1, 2, 3, 4, 5} i B = {4, 5, 6, 7, 8} . Oblicz i wypisz:
Wymagania:
Sumę zbiorów
Przecięcie
Różnicę A - B
Różnicę symetryczną
(proste)
Użyj operatorów zbiorowych ( | , & , - , ^ )
Wypisz każdy wynik z opisem

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

suma = A | B
print(f"suma: {suma}")

#przeciecie - wypisanie z obu zbiorów (elementy wspólne)
przeciecie = A & B
print(f"przeciecie: {przeciecie}")

#elementy, które są w A, ale nie ma ich w B
roznica = A - B
print(f"roznica: {roznica}")

#roznica symetryczna - elementy, które są tylko w jednym zbiorze
A_xor_B = A ^ B
print(f"roznica symetryczna: {A_xor_B}")
```
