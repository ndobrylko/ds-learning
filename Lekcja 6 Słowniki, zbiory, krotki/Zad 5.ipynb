Masz listę z duplikatami: [1, 2, 3, 2, 4, 1, 5, 3, 6, 4, 7] . Usuń duplikaty używając
zbioru i przekonwertuj z powrotem na listę. Wypisz obie listy i porównaj długości.
Wymagania:
(proste)
Użyj set() do usunięcia duplikatów
Wypisz przed i po

```python
dane = [1, 2, 3, 2, 4, 1, 5, 3, 6, 4, 7]

#aby usunąć duplikaty wystarczy zmienić listę na zbiory SET
zbior = set(dane)

#teraz chcemy spowrotem uzyskać listę jak w poleceniu
dane_bez_duplikatow = list(zbior)

print(f"Dane przed zmianą: {dane}")
print(f"Dług oryginalnej listy:", len(dane))
print(f"Dane bez duplikatów: {dane_bez_duplikatow}")
print(f"Dług listy bez duplikatów:", len(dane_bez_duplikatow))


```
