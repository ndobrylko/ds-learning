Masz słownik mapujący nazwy produktów na kategorie: {'laptop': 'elektronika',
'klawiatura': 'elektronika', 'krzesło': 'meble', 'biurko': 'meble'} . Stwórz
odwrotny słownik gdzie klucze to kategorie, a wartości to listy produktów.
Wymagania:
Oczekiwany wynik:
(średnie)
Iteruj po oryginalnym słowniku
Zgrupuj produkty według kategorii
Wynik: {'elektronika': ['laptop', 'klawiatura'], 'meble': [...]}
Oczekiwany wynik:
Słownik kategoria: lista_produktów

```python
produkty = {'laptop': 'elektronika',
'klawiatura': 'elektronika', 'krzesło': 'meble', 'biurko': 'meble'}

odwrotny = {}
for produkt, kategoria in produkty.items():
    if kategoria in odwrotny:
        odwrotny[kategoria].append(produkt)
    else:
        odwrotny[kategoria] = [produkt]

print(odwrotny)
```

```python
for produkt, kategoria in produkty.items():
    # Python pobiera każdą krotkę (klucz, wartość)
    # Rozpakowuje ją w zmienne 'produkt' i 'kategoria'
    print(produkt, kategoria)

Ważne rzeczy do zapamiętania
Python nie zgaduje, że pierwszy element to „produkt” – to Ty w kodzie nadajesz nazwę zmiennej (produkt, kategoria) = nazwy, które przypisują wartości z krotki

Liczba zmiennych po for musi dokładnie odpowiadać liczbie elementów krotki
.items() → zawsze krotka 2-elementowa (klucz, wartość)
.keys() → tylko jeden element → for klucz in produkty.keys():
```
