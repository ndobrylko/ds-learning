```python
stany = ["A", "B", "A", "A", "C", "B", "C", "A", "B", "B"]
unikalne_stany = set(stany)

licznik_przejsc = {}

for stan in stany:
    licznik_przejsc[stan] = {}

for i in range(len(stany) - 1):
    aktualny = stany[i]
    nastepny = stany[i + 1]

    if nastepny not in licznik_przejsc[aktualny]:
        licznik_przejsc[aktualny][nastepny] = 0
        licznik_przejsc[aktualny][nastepny] += 1

print(licznik_przejsc)
stany = sorted(licznik_przejsc.keys())
print("\t" + "\t".join(stany))
for aktualny in stany:
    wiersz = [str(licznik_przejsc[aktualny].get(nastepny, 0)) for nastepny in stany]
    print(aktualny + "\t" + "\t".join(wiersz))


```
