```python
def rekurencja(n):
    if n == 1:
        return 1
    else:
        return n * rekurencja(n-1)

rekurencja(4)


def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

lista_liczb = [fibonacci(x) for x in range(1,16)]
lista liczb
```
