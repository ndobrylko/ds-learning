```python
def powitanie(imie, jezyk = "polski"):
    if jezyk == "polski":
        print(f"Cześć, {imie}!")
    elif jezyk == "angielski":
        print(f"Hello, {imie}!")
    elif jezyk == "niemiecki":
        print(f"Hallo, {imie}!")

powitanie("Anna")
powitanie("Anna", "angielski")
powitanie("Anna", "niemiecki")
```
