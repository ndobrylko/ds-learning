```python
class Produkt:
    def __inint__(self, nazwa, cena, ilosc_w_magazynie):
        self.nazwa = nazwa
        self.cena = cena
        self.ilosc_w_magazynie = ilosc_w_magazynie

    def pokaz_info(self):
        print(f"nazwa: {self.nazwa}")
        print(f"cena: {self.cena} zł")
        print(f"ilosc_w_magazynie: {self.ilosc_w_magazynie}")

    def sprzedaj(self, ilosc):
        self.ilosc_w_magazynie -= ilosc
        print(f"Ilość w magazynie: {self.ilosc_w_magazynie}")

    def dostawa(self, ilosc):
        self.ilosc_w_magazynie += ilosc
        print(f"Ilość w magazynie: {self.ilosc_w_magazynie}")

my_produkt = Produkt("dd", cena:45.5, ilosc_w_magazynie: 50)
my_produkt.sprzedaj(5)
my_produkt.pokaz_info()
```
