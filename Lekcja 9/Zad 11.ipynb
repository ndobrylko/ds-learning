Zadanie 11 – System ocen uczniów
Stwórz klasy:<br>
  - Uczen (imie, oceny=[])<br>
  - Klasa (nazwa, uczniowie=[])<br>
Metody Klasa :<br>
  - dodaj_ucznia(uczen)<br>
  - srednia_klasy() – średnia ocen wszystkich uczniów<br>
  - najlepszy_uczen() – zwraca ucznia z najwyższą średnią<br>
Stwórz klasę z 5 uczniami i oblicz statystyki.
(średnie)

```python
class Uczen:
    def __init__(self, imie, oceny):
        self.imie = imie
        self.oceny = oceny

    def srednia(self):
        return sum(self.oceny)/len(self.oceny)

class Klasa:
    def __init__(self, nazwa):
        self.nazwa = nazwa
        self.uczniowie = []

    def dodaj_ucznia(self, uczeń):
        self.uczniowie.append(uczeń)

    def srednia_klasy(self):
        wszystkie_oceny = []

        for uczen in self.uczniowie:
            wszystkie_oceny.extend(uczen.oceny)
#extend() dodaje wiele elementów osobno
        return sum(wszystkie_oceny) / len(wszystkie_oceny)

    def najlepszy_uczen(self):
        return max(self.uczniowie, key=lambda uczen: uczen.srednia())

u1 = Uczen("Anna", [5, 4])
u2 = Uczen("Bartek", [3, 2])
u3 = Uczen("Kasia", [4, 4])
u4 = Uczen("Dawid", [3, 5])
u5 = Uczen("Ewa", [3, 4])

klasa_1A = Klasa("1A")

for u in [u1, u2, u3, u4, u5]:
    klasa_1A.dodaj_ucznia(u)

print(f"Średnia klasy {klasa_1A.nazwa}: {klasa_1A.srednia_klasy():.2f}")

najlepszy_u = klasa_1A.najlepszy_uczen()

if najlepszy_u:
    print(f"Najlepszy uczeń: {najlepszy_u.imie}, średnia: {najlepszy_u.srednia():.2f}")
```
