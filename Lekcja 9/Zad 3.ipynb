Stwórz klasę BankAccount z atrybutami: wlasciciel , saldo .
Metody:
(proste)
-wplac(kwota) – zwiększa saldo
-wyplac(kwota) – zmniejsza saldo (tylko jeśli wystarczy środków)
-pokaz_saldo() – wyświetla saldo
Stwórz kilka kont i przetestuj operacje.

```python
class BankAccount:
    def __init__(self, wlasciciel, saldo):
        self.wlasciciel = wlasciciel
        self.saldo = saldo

    def wplac_kwota(self, kwota):
        self.saldo += kwota
        print(f"Wpłacono {kwota}")

    def wyplac_kwota(self, kwota):
        if kwota <= self.saldo:
            self.saldo -= kwota
            print(f"Wyplacono {kwota}")
        else:
            print("Brak wystarczajacych srodkow")

    def pokaz_saldo(self):
            print(f"Konto {self.wlasciciel}: {self.saldo}")

konto1 = BankAccount("JanK", 5000)
konto2 = BankAccount("AnnaK", 8000)

konto1.pokaz_saldo()
konto1.wplac_kwota(1000)
konto1.wyplac_kwota(300)
konto1.pokaz_saldo()
print(f"------------------------------")
konto2.pokaz_saldo()
konto2.wplac_kwota(500)
konto2.wyplac_kwota(9000)
konto2.pokaz_saldo()


```
