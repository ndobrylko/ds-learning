```python
import numpy as np
import pandas as pd

from tensorflow.keras.datasets import mnist

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

import umap
import plotly.express as px
```

```python
#spłaszczanie obrazków(zamiana z 28x28 na wektor 784 liczb)
X = X_train.reshape(X_train.shape[0], -1)
y = y_train

print(X.shape)
print(y.shape)
```

```python
#bierzemy próbkę
np.random.seed(42)

n_samples = 3000
idx = np.random.choice(len(X), n_samples, replace=False)

X_sample = X[idx]
y_sample = y[idx]

print(X_sample.shape)
print(y_sample.shape)
```

```python
#skalowanie danych
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_sample)
```

```python
#PCA 2D
pca_2d = PCA(n_components=2, random_state=42)
X_pca_2d = pca_2d.fit_transform(X_scaled)

print(X_pca_2d.shape)
print("Explained variance:", pca_2d.explained_variance_ratio_.sum())
```

```python
#wykres interaktywny PCA 2D
df_pca_2d = pd.DataFrame({
    "x": X_pca_2d[:, 0],
    "y": X_pca_2d[:, 1],
    "label": y_sample.astype(str)
})

fig = px.scatter(
    df_pca_2d,
    x="x",
    y="y",
    color="label",
    title="PCA 2D — MNIST",
    opacity=0.7
)

fig.show()
```

```python
#PCA 3D
pca_3d = PCA(n_components=3, random_state=42)
X_pca_3d = pca_3d.fit_transform(X_scaled)

print(X_pca_3d.shape)
print("Explained variance:", pca_3d.explained_variance_ratio_.sum())
```

```python
df_pca_3d = pd.DataFrame({
    "x": X_pca_3d[:, 0],
    "y": X_pca_3d[:, 1],
    "z": X_pca_3d[:, 2],
    "label": y_sample.astype(str)
})

fig = px.scatter_3d(
    df_pca_3d,
    x="x",
    y="y",
    z="z",
    color="label",
    title="PCA 3D — MNIST",
    opacity=0.7
)

fig.show()
```

```python
#t-SNE 2D
tsne_2d = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate="auto",
    init="pca",
    random_state=42
)

X_tsne_2d = tsne_2d.fit_transform(X_scaled)

print(X_tsne_2d.shape)
```

```python
df_tsne_2d = pd.DataFrame({
    "x": X_tsne_2d[:, 0],
    "y": X_tsne_2d[:, 1],
    "label": y_sample.astype(str)
})

fig = px.scatter(
    df_tsne_2d,
    x="x",
    y="y",
    color="label",
    title="t-SNE 2D — MNIST",
    opacity=0.7
)

fig.show()
```

```python
#t-SNE 3D
tsne_3d = TSNE(
    n_components=3,
    perplexity=30,
    learning_rate="auto",
    init="pca",
    random_state=42
)

X_tsne_3d = tsne_3d.fit_transform(X_scaled)

print(X_tsne_3d.shape)
```

```python
df_tsne_3d = pd.DataFrame({
    "x": X_tsne_3d[:, 0],
    "y": X_tsne_3d[:, 1],
    "z": X_tsne_3d[:, 2],
    "label": y_sample.astype(str)
})

fig = px.scatter_3d(
    df_tsne_3d,
    x="x",
    y="y",
    z="z",
    color="label",
    title="t-SNE 3D — MNIST",
    opacity=0.7
)

fig.show()
```

```python
#UMAP 2D
umap_2d = umap.UMAP(
    n_components=2,
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap_2d = umap_2d.fit_transform(X_scaled)

print(X_umap_2d.shape)
```

```python
df_umap_2d = pd.DataFrame({
    "x": X_umap_2d[:, 0],
    "y": X_umap_2d[:, 1],
    "label": y_sample.astype(str)
})

fig = px.scatter(
    df_umap_2d,
    x="x",
    y="y",
    color="label",
    title="UMAP 2D — MNIST",
    opacity=0.7
)

fig.show()
```

```python
#UMAP 3D
umap_3d = umap.UMAP(
    n_components=3,
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap_3d = umap_3d.fit_transform(X_scaled)

print(X_umap_3d.shape)
```

```python
df_umap_3d = pd.DataFrame({
    "x": X_umap_3d[:, 0],
    "y": X_umap_3d[:, 1],
    "z": X_umap_3d[:, 2],
    "label": y_sample.astype(str)
})

fig = px.scatter_3d(
    df_umap_3d,
    x="x",
    y="y",
    z="z",
    color="label",
    title="UMAP 3D — MNIST",
    opacity=0.7
)

fig.show()
```

PCA jest metodą liniową, dlatego na MNIST pokazuje o**gólną strukturę danych, ale klastry cyfr częściowo się nakładają**. Szybkie i dobre do preprocessingu.

t-SNE daje znacznie **lepszą separację** lokalnych grup, dlatego cyfry tworzą bardziej wyraźne klastry. Metoda nadaje się **głównie do wizualizacji**, a nie jako preprocessing do modeli ML.

UMAP również **dobrze separuje cyfry, często szybciej niż t-SNE**. Dodatkowo lepiej zachowuje część struktury globalnej i **może być używany także jako preprocessing**.

Wizualizacje 3D pozwalają lepiej zobaczyć separację niektórych klas, ponieważ można obracać wykres i obserwować klastry z różnych perspektyw.
